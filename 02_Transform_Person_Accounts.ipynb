```python
# bring in emails to the contacts df

# Isolate the necessary columns and rename 'address' in flight
emails_subset = df_emails_clean[['emailable_id', 'address']].rename(columns={'address': 'email_address'})

# Left join to the contacts dataframe
df_contacts_clean = df_contacts_clean.merge(
    emails_subset, 
    left_on='id', 
    right_on='emailable_id', 
    how='left'
)

# Drop the right-side join key to keep the dataframe clean
df_contacts_clean = df_contacts_clean.drop(columns=['emailable_id'])
```

```python
# Join addresses to the contacts dataframe

# 1. Safely concatenate the street address fields first
df_addresses_clean['street_address'] = df_addresses_clean['street_address'].fillna('')
df_addresses_clean['secondary_address'] = df_addresses_clean['secondary_address'].fillna('')

df_addresses_clean['full_street'] = (
    df_addresses_clean['street_address'].astype(str) + ' ' + df_addresses_clean['secondary_address'].astype(str)
).str.strip()

# 2. Isolate core columns for pivoting
addr_cols = [
    'addressable_id', 
    'full_street', 
    'city', 
    'state', 
    'zip', 
    'is_primary'
]
addr_subset = df_addresses_clean[addr_cols].copy()

# 3. Sort by contact ID and is_primary (True values bubble to the top)
# If a secondary sort criteria exists (like 'last_updated'), add it to the list to break ties deterministically
addr_subset = addr_subset.sort_values(by=['addressable_id', 'is_primary'], ascending=[True, False])

# 4. Rank addresses per contact (0-indexed)
addr_subset['address_rank'] = addr_subset.groupby('addressable_id').cumcount()

# Filter to keep only the top 2 (rank 0 = Primary, rank 1 = Secondary)
addr_subset = addr_subset[addr_subset['address_rank'] < 2]

# Map the numeric rank to column prefixes for the pivot
addr_subset['prefix'] = addr_subset['address_rank'].map({0: 'Primary', 1: 'Secondary'})
addr_subset = addr_subset.drop(columns=['address_rank', 'is_primary'])

# 5. Pivot from Long to Wide format
addr_wide = addr_subset.pivot(index='addressable_id', columns='prefix')

# Flatten the MultiIndex columns created by the pivot (e.g., changes ('city', 'Primary') to 'Primary_city')
addr_wide.columns = [f"{prefix}_{col}" for col, prefix in addr_wide.columns]
addr_wide = addr_wide.reset_index()

# 6. Execute the scalable left merge
df_contacts_clean = df_contacts_clean.merge(
    addr_wide,
    left_on='id',
    right_on='addressable_id',
    how='left'
)

# 7. Clean up the redundant join key
df_contacts_clean = df_contacts_clean.drop(columns=['addressable_id'])
```

```python
# Bring in phones to the contacts df

import pandas as pd
import re

# 1. Isolate the necessary columns and rename 'number' in flight
# Assuming 'is_primary' is the boolean flag in your phones table
phone_cols = ['callable_id', 'number', 'is_primary']
phones_subset = df_phones_clean[phone_cols].rename(columns={'number': 'phone_number'}).copy()

# 2. Sanitize the phone numbers (Optional but highly recommended)
# This strips everything except digits, 'x', and '+' (to preserve extensions and international codes)
phones_subset['phone_number'] = phones_subset['phone_number'].astype(str).apply(
    lambda x: re.sub(r'[^\d+xX]', '', x) if pd.notnull(x) else x
)

# 3. Sort by contact ID and is_primary (True values bubble to the top)
phones_subset = phones_subset.sort_values(by=['callable_id', 'is_primary'], ascending=[True, False])

# 4. Rank phones per contact (0-indexed)
phones_subset['phone_rank'] = phones_subset.groupby('callable_id').cumcount()

# Filter to keep only the top 2 (Primary and Secondary)
# If your target Salesforce schema only accepts a single phone, change this to `< 1`
phones_subset = phones_subset[phones_subset['phone_rank'] < 2]

# Map the numeric rank to column prefixes for the pivot
phones_subset['prefix'] = phones_subset['phone_rank'].map({0: 'Primary', 1: 'Secondary'})
phones_subset = phones_subset.drop(columns=['phone_rank', 'is_primary'])

# 5. Pivot from Long to Wide format
phones_wide = phones_subset.pivot(index='callable_id', columns='prefix')

# Flatten the MultiIndex columns created by the pivot (e.g., ('phone_number', 'Primary') to 'Primary_phone_number')
phones_wide.columns = [f"{prefix}_{col}" for col, prefix in phones_wide.columns]
phones_wide = phones_wide.reset_index()

# 6. Execute the scalable left merge
df_contacts_clean = df_contacts_clean.merge(
    phones_wide,
    left_on='id',
    right_on='callable_id',
    how='left'
)

# 7. Clean up the redundant join key
df_contacts_clean = df_contacts_clean.drop(columns=['callable_id'])
```

```python
# Bring in important information to the contacts df

# 1. Isolate the necessary columns and rename 'content' to 'Description'
info_cols = ['contact_id', 'content']
info_subset = df_contact_important_informations_clean[info_cols].rename(columns={'content': 'Description'})

# 2. Defensively group and aggregate strings 
# This ensures that if a contact accidentally has multiple important info rows, 
# they are combined with line breaks rather than duplicating the contact row during the merge.
info_aggregated = (
    info_subset.groupby('contact_id')['Description']
    .apply(lambda x: '\n\n'.join(x.dropna().astype(str)))
    .reset_index()
)

# 3. Execute the scalable left merge
df_contacts_clean = df_contacts_clean.merge(
    info_aggregated,
    left_on='id',
    right_on='contact_id',
    how='left'
)

# 4. Clean up the redundant join key
df_contacts_clean = df_contacts_clean.drop(columns=['contact_id'])
```

```python
# Bring in KYC information to the contacts df

import pandas as pd

# 1. Isolate and rename columns to prevent naming collisions (e.g., 'state')
kyc_cols = ['contact_id', 'number', 'state', 'expiration_date']
kyc_subset = df_contact_identifications_clean[kyc_cols].rename(columns={
    'number': 'kyc_id_number',
    'state': 'kyc_id_state',
    'expiration_date': 'kyc_expiration_date'
}).copy()

# 2. Defensively sort to prioritize the most current identification
# Convert to datetime first to ensure accurate sorting, coercing bad legacy strings to NaT
kyc_subset['kyc_expiration_date'] = pd.to_datetime(kyc_subset['kyc_expiration_date'], errors='coerce')

# Sort so that the newest expiration dates are at the top
kyc_subset = kyc_subset.sort_values(by=['contact_id', 'kyc_expiration_date'], ascending=[True, False])

# Drop duplicates to maintain a strict 1:1 relationship for the Salesforce payload
kyc_subset = kyc_subset.drop_duplicates(subset=['contact_id'], keep='first')

# Convert datetime back to a standardized string format for Data Loader (YYYY-MM-DD)
kyc_subset['kyc_expiration_date'] = kyc_subset['kyc_expiration_date'].dt.strftime('%Y-%m-%d')

# 3. Execute the scalable left merge
df_contacts_clean = df_contacts_clean.merge(
    kyc_subset,
    left_on='id',
    right_on='contact_id',
    how='left'
)

# 4. Clean up the redundant join key
df_contacts_clean = df_contacts_clean.drop(columns=['contact_id'])
```

```python
# Death handling

import pandas as pd

# 1. Create a boolean mask to identify deceased individuals.
# We check for notna() and also ensure it isn't an empty string just in case 
# the legacy CSV extraction left blanks instead of true nulls.
is_deceased_mask = (
    (df_contacts_clean['death_date'].notna() & (df_contacts_clean['death_date'] != '')) | 
    (df_contacts_clean['status'] == 'Deceased')
)

# 2. Initialize the Salesforce custom field as False for everyone by default
df_contacts_clean['Deceased__c'] = False

# 3. Apply the True flag where the mask conditions are met
df_contacts_clean.loc[is_deceased_mask, 'Deceased__c'] = True

# 4. Override the status to 'Active Client' for those same records
df_contacts_clean.loc[is_deceased_mask, 'status'] = 'Active Client'
```

```python
from os import path

import pandas as pd
from openpyxl.styles import PatternFill, Font, Border, Side, Alignment
from openpyxl.utils import get_column_letter

# Define your output paths (you can parameterize these with your M&A prefix if needed)
#upload_path = 'salesforce_account_export.csv'
#raw_data_path = 'redtail_migration_raw_data.xlsx'

# --- 1. Export the CSV Payload for Data Loader ---
#df_contacts_export.to_csv(path(upload_path / '02_Transform_Person_Accounts.csv'), index=False)


# --- 2. Export the Audit Trail Excel Workbook ---
dfs = {
    'Export Payload': df_contacts_export,
    'Status': df_statuses,
    'Phones Raw': df_phones,
    'Phones Clean': df_phones_clean,
    'Contacts Raw': df_contacts,
    'Contacts Clean': df_contacts_clean,
    'Emails Raw': df_emails,
    'Emails Clean': df_emails_clean,
    'Contact Important Info Raw': df_contact_important_informations,
    'Contact Important Info Clean': df_contact_important_informations_clean,
    'Contact Identifications Raw': df_contact_identifications,
    'Contact Identifications Clean': df_contact_identifications_clean,
    'Addressess Raw': df_addresses,
    'Addressess Clean': df_addresses_clean
}

with pd.ExcelWriter(excel_file, engine='openpyxl') as writer:
    
    # Pre-define Excel styling properties
    header_fill = PatternFill(start_color="2E3440", end_color="2E3440", fill_type="solid")
    header_font = Font(color="ECEFF4", bold=True)
    border_bottom = Border(bottom=Side(style='thin', color='D8DEE9'))
    row_fill_alt = PatternFill(start_color="F9F9FB", end_color="F9F9FB", fill_type="solid")
    
    for sheet_name, df in dfs.items():
        # Make a copy to avoid SettingWithCopy warnings during formatting
        df_out = df.copy()
        
        # Defensive check: Drop timezone info from any datetime columns to prevent Excel export errors
        dt_cols = df_out.select_dtypes(include=['datetime64', 'datetimetz']).columns
        for col in dt_cols:
            df_out[col] = df_out[col].dt.tz_localize(None)
            
        # Write the dataframe to its designated sheet
        df_out.to_excel(writer, sheet_name=sheet_name, index=False)
        worksheet = writer.sheets[sheet_name]
        
        # Apply formatting column by column
        for col_num, col_name in enumerate(df_out.columns.values):
            col_letter = get_column_letter(col_num + 1)
            
            # Header styling
            header_cell = worksheet[f"{col_letter}1"]
            header_cell.fill = header_fill
            header_cell.font = header_font
            header_cell.alignment = Alignment(horizontal='center', vertical='center')
            header_cell.border = border_bottom
            
            # Auto-adjust column width (capped at 40 width)
            if not df_out[col_name].empty:
                # 1. Use .str.len() instead of .map(len) for safe handling of nulls/ExtensionArrays
                max_data_len = df_out[col_name].astype(str).str.len().max()
                
                # 2. If the column is completely null, max_data_len becomes NaN. Catch it here.
                if pd.isna(max_data_len):
                    max_data_len = 0
                    
                # 3. Compare data length to column header length
                max_length = max(int(max_data_len), len(str(col_name)))
            else:
                max_length = len(str(col_name))
            
            worksheet.column_dimensions[col_letter].width = min(max_length + 2, 40)
            
        # Freeze the top row for auditing
        worksheet.freeze_panes = "A2"
        
        # Apply subtle zebra striping
        for row_num in range(2, worksheet.max_row + 1):
            if row_num % 2 == 0:
                for col_num in range(1, worksheet.max_column + 1):
                    worksheet.cell(row=row_num, column=col_num).fill = row_fill_alt

#print(f"Pipeline checkpoint saved successfully:\n - CSV: {csv_file}\n - Excel Audit: {excel_file}")
```
