```python
import pandas as pd
from pathlib import Path

# Setup paths (adjust these if running in a separate notebook)
file_path = Path('G:/My Drive/KM Jada Files')
upload_path = file_path / "Uploads"

# 1. Load the raw files
df_w = pd.read_csv('df_workflows.csv')
df_s = pd.read_csv('df_workflow_steps.csv')
df_t = pd.read_csv('df_workflow_tasks.csv')

# 2. Hard filter out deleted records
df_w = df_w[df_w['deleted'] == 0]
df_s = df_s[df_s['deleted'] == 0]
df_t = df_t[df_t['deleted'] == 0]

# --- PART A: BUILD THE PARENT CASES (WORKFLOWS) ---
# Ensure you query SF for the correct Record Type ID for Workflow Cases
WORKFLOW_CASE_RT_ID = '012_YOUR_WORKFLOW_CASE_RT_ID' 
MA_PREFIX = "JJ_RT_202611"

df_cases_export = pd.DataFrame({
    'RecordTypeId': WORKFLOW_CASE_RT_ID,
    'Subject': df_w['name'],
    'Description': df_w['description'].fillna(''),
    # Map status integers to your SF Case Status picklist values
    'Status': df_w['status'].map({1: 'Open', 2: 'Completed'}).fillna('Open'),
    'Origin': '', # Leaving blank per standard
    'FinServ__SourceSystemId__c': df_w['id'].apply(lambda x: f"{MA_PREFIX}_WF_{x}"),
    
    # ACTION REQUIRED: You must merge your Redtail 'workflow_links' table against df_w 
    # to find the Contact ID, then crosswalk that to the Household External ID.
    'AccountId_External__c': 'PENDING_WORKFLOW_LINKS_MERGE' 
})


# --- PART B: BUILD THE CHILD TASKS (STEPS + TASKS) ---
# 1. Join Tasks up to Steps to inherit the parent workflow_id and step name
df_tasks_flattened = df_t.merge(
    df_s[['id', 'name', 'workflow_id']], 
    left_on='workflow_step_id', 
    right_on='id', 
    how='inner',
    suffixes=('_task', '_step')
)

# 2. Format the unified Task export
df_tasks_export = pd.DataFrame({
    # Combines Step Name + Task Name for context (e.g. "Gather Docs - Send eSignature")
    'Subject': df_tasks_flattened['name_step'].astype(str) + " - " + df_tasks_flattened['name_task'].astype(str),
    'Description': df_tasks_flattened['description'].fillna(''),
    
    # If completion_date is populated, mark Completed, else Open
    'Status': df_tasks_flattened['completion_date'].apply(lambda x: 'Completed' if pd.notna(x) else 'Open'),
    'ActivityDate': pd.to_datetime(df_tasks_flattened['due_date']).dt.strftime('%Y-%m-%d'),
    
    # Generate unique Task ID
    'FinServ__SourceSystemId__c': df_tasks_flattened['id_task'].apply(lambda x: f"{MA_PREFIX}_WFT_{x}"),
    
    # Crucial Relational Link: Point the WhatId to the newly generated Workflow Case
    'WhatId_External__c': df_tasks_flattened['workflow_id'].apply(lambda x: f"{MA_PREFIX}_WF_{x}")
})


# --- PART C: EXPORT FOR DATA LOADER ---
df_cases_export.to_csv(upload_path / '05A_Transform_Workflow_Cases.csv', index=False)
df_tasks_export.to_csv(upload_path / '05B_Transform_Workflow_Tasks.csv', index=False)

print(f"Generated {len(df_cases_export)} Parent Workflow Cases.")
print(f"Generated {len(df_tasks_export)} Child Workflow Tasks.")
```
