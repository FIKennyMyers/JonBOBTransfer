```python
import pandas as pd
from pathlib import Path
from openpyxl.styles import PatternFill, Font, Border, Side, Alignment
from openpyxl.utils import get_column_letter



# 1. Load the raw workflow CSVs and Person Account export
df_w = df_workflows
df_s = df_workflow_steps
df_t = df_workflow_tasks
df_cw = df_contact_workflows

df_pa = pd.read_csv(upload_path / '02_Transform_Person_Accounts.csv')

# 2. Filter active records (preserving raw DataFrames for the compliance workbook)
df_w_clean = df_w[df_w['deleted'] == 0].copy()
df_s_clean = df_s[df_s['deleted'] == 0].copy()
df_t_clean = df_t[df_t['deleted'] == 0].copy()

# Deduplicate junction table to 1 primary contact per workflow (prevents Case fan-out)
df_cw_clean = (
    df_cw[df_cw['deleted'] == 0]
    .sort_values('id')
    .drop_duplicates(subset=['workflow_id'])
    .copy()
)

# 3. Build the Workflow -> Contact Crosswalk
df_pa_sub = (
    df_pa[['id', 'FinServ__SourceSystemId__c']]
    .dropna(subset=['id'])
    .astype({'id': 'int64'})
    .drop_duplicates(subset=['id'])
)

df_wf_crosswalk = df_cw_clean.merge(
    df_pa_sub,
    left_on='contact_id',
    right_on='id',
    how='inner',
    suffixes=('_link', '_pa')
).rename(columns={'FinServ__SourceSystemId__c': 'Contact_Ext_ID'})

# Attach Contact_Ext_ID to the active workflows
df_w_mapped = df_w_clean.merge(
    df_wf_crosswalk[['workflow_id', 'contact_id', 'Contact_Ext_ID']],
    left_on='id',
    right_on='workflow_id',
    how='inner'
)

# --- PART A: BUILD THE PARENT CASES (WORKFLOWS) ---
WORKFLOW_CASE_RT_ID = '012_YOUR_WORKFLOW_CASE_RT_ID'
MA_PREFIX = "JJ_RT_202611"

df_cases_export = pd.DataFrame({
    'RecordTypeId': WORKFLOW_CASE_RT_ID,
    'Subject': df_w_mapped['name'],
    'Description': df_w_mapped['description'].fillna(''),
    'Status': df_w_mapped['status'].map({1: 'Open', 2: 'Complete'}).fillna('Open'),
    'Origin': '',
    'FinServ__SourceSystemId__c': df_w_mapped['id'].apply(lambda x: f"{MA_PREFIX}_WF_{x}"),
    'ContactId_External__c': df_w_mapped['Contact_Ext_ID']
})

# --- PART B: BUILD THE CHILD TASKS (STEPS + TASKS) ---
df_tasks_flattened = df_t_clean.merge(
    df_s_clean[['id', 'name', 'workflow_id']],
    left_on='workflow_step_id',
    right_on='id',
    how='inner',
    suffixes=('_task', '_step')
)

# Keep only tasks whose parent workflow survived the active contact mapping
df_tasks_flattened = df_tasks_flattened[
    df_tasks_flattened['workflow_id'].isin(df_w_mapped['id'])
].copy()

df_tasks_export = pd.DataFrame({
    'Subject': df_tasks_flattened['name_step'].astype(str) + " - " + df_tasks_flattened['name_task'].astype(str),
    'Description': df_tasks_flattened['description'].fillna(''),
    'Status': df_tasks_flattened['completion_date'].apply(lambda x: 'Completed' if pd.notna(x) else 'Open'),
    'ActivityDate': pd.to_datetime(df_tasks_flattened['due_date'], errors='coerce').dt.strftime('%Y-%m-%d'),
    'FinServ__SourceSystemId__c': df_tasks_flattened['id_task'].apply(lambda x: f"{MA_PREFIX}_WFT_{x}"),
    'WhatId_External__c': df_tasks_flattened['workflow_id'].apply(lambda x: f"{MA_PREFIX}_WF_{x}")
})

# --- 1. Export the CSV Payloads for Data Loader ---
cases_csv = upload_path / '05A_Transform_Workflow_Cases.csv'
tasks_csv = upload_path / '05B_Transform_Workflow_Tasks.csv'

df_cases_export.to_csv(cases_csv, index=False)
df_tasks_export.to_csv(tasks_csv, index=False)

# --- 2. Export the Audit Trail Excel Workbook for Compliance ---
excel_file = raw_data_path / '05_transform_workflows_raw.xlsx'

dfs = {
    'Export Cases Payload': df_cases_export,
    'Export Tasks Payload': df_tasks_export,
    'Workflows Mapped': df_w_mapped,
    'Tasks Flattened': df_tasks_flattened,
    'Contact Workflows Clean': df_cw_clean,
    'Workflows Raw': df_w,
    'Workflow Steps Raw': df_s,
    'Workflow Tasks Raw': df_t,
    'Contact Workflows Raw': df_cw
}

with pd.ExcelWriter(excel_file, engine='openpyxl') as writer:
    
    # Pre-define Excel styling properties
    header_fill = PatternFill(start_color="2E3440", end_color="2E3440", fill_type="solid")
    header_font = Font(color="ECEFF4", bold=True)
    border_bottom = Border(bottom=Side(style='thin', color='D8DEE9'))
    row_fill_alt = PatternFill(start_color="F9F9FB", end_color="F9F9FB", fill_type="solid")
    
    for sheet_name, df in dfs.items():
        df_out = df.copy()
        
        # Defensive check: Drop timezone info from any datetime columns to prevent Excel export errors
        dt_cols = df_out.select_dtypes(include=['datetime64', 'datetimetz']).columns
        for col in dt_cols:
            df_out[col] = df_out[col].dt.tz_localize(None)
            
        # Defensive truncation for Excel's 32,767 cell character limit
        for col in df_out.select_dtypes(include=['object', 'string']).columns:
            df_out[col] = df_out[col].fillna('').astype(str).str[:32000]
            
        # Write the dataframe to its designated sheet
        df_out.to_excel(writer, sheet_name=sheet_name, index=False)
        worksheet = writer.sheets[sheet_name]
        
        # Apply formatting column by column
        for col_num, col_name in enumerate(df_out.columns.values):
            col_letter = get_column_letter(col_num + 1)
            
            # Header styling
            header_cell = worksheet[f"{col_letter}1"]
            header_cell.fill = header_fill
            header_cell.font = header_font
            header_cell.alignment = Alignment(horizontal='center', vertical='center')
            header_cell.border = border_bottom
            
            # Auto-adjust column width (capped at 40 width)
            if not df_out[col_name].empty:
                max_data_len = df_out[col_name].astype(str).str.len().max()
                if pd.isna(max_data_len):
                    max_data_len = 0
                max_length = max(int(max_data_len), len(str(col_name)))
            else:
                max_length = len(str(col_name))
            
            worksheet.column_dimensions[col_letter].width = min(max_length + 2, 40)
            
        # Freeze the top row for auditing
        worksheet.freeze_panes = "A2"
        
        # Apply subtle zebra striping
        for row_num in range(2, worksheet.max_row + 1):
            if row_num % 2 == 0:
                for col_num in range(1, worksheet.max_column + 1):
                    worksheet.cell(row=row_num, column=col_num).fill = row_fill_alt

print(f"Pipeline checkpoint saved successfully:\n - Cases CSV: {cases_csv} ({len(df_cases_export)} rows)\n - Tasks CSV: {tasks_csv} ({len(df_tasks_export)} rows)\n - Excel Audit: {excel_file}")
```
