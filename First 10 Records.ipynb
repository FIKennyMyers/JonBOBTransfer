```python
import pandas as pd
from sqlalchemy import create_engine
import urllib.parse
from getpass import getpass

# 1. Gather your credentials (mirroring your MySQL Workbench setup)
DB_USER = "kenny.myers"          # Username from Workbench
DB_HOST = "127.0.0.1"   # The Host/IP from Workbench
DB_PORT = "3307"                   # Default MySQL port
DB_NAME = "newschema"              # Database name from your Redtail header

# 2. Securely prompt for password
DB_PASS = getpass("Enter Database Password: ")

# 3. Construct the SQLAlchemy Connection URI
# Format: mysql+pymysql://user:pass@host:port/dbname
connection_uri = f"mysql+pymysql://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}/{DB_NAME}"

# Create the SQLAlchemy engine
engine = create_engine(connection_uri)

# The target tables provided for the data dictionary
tables = [
    "addresses", "contact_advisor_details", "contacts", "contact_sams",
    "contact_important_informations", "contact_recruits", "contact_wars",
    "contact_taxes", "contact_personal_profiles", "contact_keywords",
    "phones", "contact_identifications", "contact_family_members",
    "emails", "contact_membership_codes", "contact_tags_members",
    "contact_families", "contact_roles", "contact_salutations",
    "contact_categories", "contact_udf_field_values", "contact_statuses",
    "professions", "quicklists", "custom_export_fields", "contact_udfs",
    "contact_urls", "note_associations", "objectives",
    "activity_completed_instances", "notes", "activity_attendees",
    "activities", "activities_contacts"
]

output_file = 'G:\\My Drive\\KM Jada Files\\Redtail_Schema_Sample.xlsx'

print("Initiating database extraction...")

# 2. Extract and write to Excel
# openpyxl is the standard engine for modern .xlsx files in Pandas
with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
    for table in tables:
        try:
            # f-strings are safe here because our table list is strictly hardcoded
            query = f"SELECT * FROM {table} LIMIT 10"
            
            # Read directly into a dataframe
            df = pd.read_sql(query, engine)
            
            # Excel imposes a strict 31-character limit on worksheet tab names.
            # We slice the string to ensure it doesn't crash the writer.
            sheet_name = table[:31] 
            
            # Write to the specific tab, dropping the Pandas integer index
            df.to_excel(writer, sheet_name=sheet_name, index=False)
            
            print(f"Success: [{table}] -> {len(df)} rows written.")
            
        except Exception as e:
            # If a table is deprecated, empty, or misnamed in your dump, 
            # this catches the error and moves to the next table without killing the script.
            print(f"Error extracting [{table}]: {e}")

print(f"\nExtraction complete. File saved to local directory as: {output_file}")
```
