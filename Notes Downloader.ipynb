```python
from bs4 import BeautifulSoup

df_notes = pd.read_sql("SELECT * FROM notes", engine)

# Clean HTML in memory
def strip_html(text):
    if pd.isna(text):
        return text
    soup = BeautifulSoup(str(text), "html.parser")
    return soup.get_text(separator=' | ', strip=True)

df_notes['body_clean'] = df_notes['body'].apply(strip_html)
```

```python
# Calculate the character count for each note safely
df_notes['body_char_count'] = df_notes['body'].astype(str).str.len()

# Isolate the records that will fail a standard Task import
df_too_long = df_notes[df_notes['body_char_count'] > 31500]
```
